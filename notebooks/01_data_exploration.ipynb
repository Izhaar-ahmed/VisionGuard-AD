# 01 — MVTec-AD Data Exploration

Explore the MVTec Anomaly Detection dataset structure, statistics, and sample images.

```python
import sys
sys.path.insert(0, '..')

import matplotlib.pyplot as plt
import numpy as np
from data.mvtec_dataset import MVTecDataset, MVTEC_CATEGORIES, IMAGENET_MEAN, IMAGENET_STD

DATA_ROOT = '../data/mvtec'
print(f'MVTec categories: {MVTEC_CATEGORIES}')
```

```python
# Dataset statistics for each category
for cat in MVTEC_CATEGORIES:
    try:
        ds = MVTecDataset(DATA_ROOT, cat, 'test', img_size=224)
        stats = ds.get_statistics()
        print(f"{cat:15s} | total={stats['total_images']:4d} | normal={stats['normal_count']:3d} | defect={stats['defect_count']:3d} | ratio={stats['class_imbalance_ratio']:.1f}")
    except Exception as e:
        print(f"{cat:15s} | ERROR: {e}")
```

```python
# Visualize samples from one category
def denorm(img_tensor):
    img = img_tensor.permute(1,2,0).numpy()
    img = img * np.array(IMAGENET_STD) + np.array(IMAGENET_MEAN)
    return np.clip(img, 0, 1)

cat = 'carpet'
try:
    ds = MVTecDataset(DATA_ROOT, cat, 'test', img_size=224)
    fig, axes = plt.subplots(2, 5, figsize=(15, 6))
    for i, ax in enumerate(axes.flat):
        if i < len(ds):
            img, mask, label, dtype, path = ds[i]
            ax.imshow(denorm(img))
            color = 'red' if label == 1 else 'green'
            ax.set_title(dtype, color=color, fontsize=9)
            ax.axis('off')
    plt.suptitle(f'{cat} — Test Samples', fontsize=14)
    plt.tight_layout()
    plt.show()
except Exception as e:
    print(f'Dataset not found: {e}')
```
