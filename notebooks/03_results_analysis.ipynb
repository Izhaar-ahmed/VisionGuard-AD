# 03 — Results Analysis

Analyze benchmark results, compare methods, and visualize performance across categories.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load benchmark results
try:
    with open('../outputs/benchmark/benchmark_results.json') as f:
        results = json.load(f)
    df = pd.DataFrame(results)
    print(df.to_string(index=False))
except FileNotFoundError:
    print('No benchmark results found. Run benchmark.py first.')
    # Create dummy data for demonstration
    df = pd.DataFrame({
        'category': ['carpet','grid','leather','tile','wood','bottle','cable','capsule','hazelnut','metal_nut','pill','screw','toothbrush','transistor','zipper'],
        'image_auroc': [98.7,99.1,100,98.5,99.2,100,97.8,98.1,100,99.5,96.3,98.2,100,99.7,98.9],
        'pixel_auroc': [98.3,97.5,99.1,94.8,95.2,98.7,97.1,98.5,98.9,97.3,96.8,99.2,98.6,96.4,98.1],
        'pro_score': [94.1,93.2,97.5,88.3,90.1,95.2,92.8,94.7,96.1,93.5,91.2,97.8,94.3,88.9,93.6],
    })
```

```python
# Bar chart comparison
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
metrics = ['image_auroc', 'pixel_auroc', 'pro_score']
titles = ['Image-Level AUROC', 'Pixel-Level AUROC', 'PRO Score']
colors = ['#2196F3', '#4CAF50', '#FF9800']

for ax, metric, title, color in zip(axes, metrics, titles, colors):
    if metric in df.columns:
        bars = ax.barh(df['category'], df[metric], color=color, alpha=0.8)
        ax.set_xlim(80, 101)
        ax.set_xlabel('%')
        ax.set_title(title, fontsize=13)
        ax.axvline(df[metric].mean(), color='red', linestyle='--', label=f'Mean: {df[metric].mean():.1f}%')
        ax.legend()
        ax.grid(axis='x', alpha=0.3)

plt.suptitle('VisionGuard-AD Benchmark Results', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Summary statistics
print('\n=== Summary ===')
for metric in metrics:
    if metric in df.columns:
        vals = df[metric]
        print(f'{metric:15s} | Mean: {vals.mean():.1f}% | Min: {vals.min():.1f}% | Max: {vals.max():.1f}% | Std: {vals.std():.1f}%')
```
