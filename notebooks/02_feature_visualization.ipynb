# 02 — Feature Visualization

Visualize intermediate CNN features extracted by the backbone, PCA of patch embeddings, and t-SNE clustering.

```python
import sys
sys.path.insert(0, '..')

import torch
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

from models.backbones.feature_extractor import FeatureExtractor, PatchEmbedding

device = 'cpu'
fe = FeatureExtractor('resnet18', device=device)
pe = PatchEmbedding(patch_size=3)
print('Feature dims:', fe.get_feature_dims())
```

```python
# Extract features from a dummy image
dummy = torch.randn(1, 3, 224, 224)
features = fe(dummy)

fig, axes = plt.subplots(1, len(features), figsize=(12, 4))
for ax, (name, feat) in zip(axes, features.items()):
    # Show mean activation across channels
    act = feat[0].mean(dim=0).cpu().numpy()
    ax.imshow(act, cmap='viridis')
    ax.set_title(f'{name} ({feat.shape[1]}ch, {feat.shape[2]}×{feat.shape[3]})')
    ax.axis('off')
plt.suptitle('Feature Map Activations', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# PCA of patch embeddings
embeddings = pe(features)
print(f'Patch embeddings shape: {embeddings.shape}')

pca = PCA(n_components=2)
emb_2d = pca.fit_transform(embeddings.detach().numpy())

plt.figure(figsize=(8, 6))
plt.scatter(emb_2d[:, 0], emb_2d[:, 1], s=1, alpha=0.5)
plt.title('PCA of Patch Embeddings')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.grid(True, alpha=0.3)
plt.show()
```
